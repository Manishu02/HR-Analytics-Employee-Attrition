## 🛠 Handling Missing Values
Imputed missing values in YearsWithCurrManager using JobRole-wise median.

```python
data['YearsWithCurrManager'] = data.groupby('JobRole')['YearsWithCurrManager']\
    .transform(lambda x: x.fillna(x.median()))
```

## 🔧 Feature Engineering

Created new features for better analysis.

```python
data['Attrition_Flag'] = data['Attrition'].map({'Yes':1, 'No':0})
```
